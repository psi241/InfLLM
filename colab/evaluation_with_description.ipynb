```python
import json
all_result = None
with open("result/result.json", "r") as f:
  # Consult on December 9, 2025 with https://docs.python.org/3/library/json.html
  texts = ""
  for line in f:
    texts += line
  all_result = json.loads(texts)


for full_name, short_name in [
    ("Qwen2-0.5B-Instruct_origin.jsonl", "origin"),
    ("Qwen2-0.5B-Instruct_inf-llm.jsonl", "inf-llm-16"),
    ("Qwen2-0.5B-Instruct_inf-llm-longer.jsonl", "inf-llm-24"),
    ("Qwen2-0.5B-Instruct_origin-rag.jsonl", "retriever-only"),
    ("Qwen2-0.5B-Instruct_inf-llm-rag.jsonl", "retrv-inf"),
]:
  all_result[short_name] = all_result[full_name]

single = ['narrativeqa', 'qasper', 'multifieldqa_en']
multiple = ['hotpotqa', '2wikimqa', 'musique']
summary = ['gov_report', 'qmsum', 'multi_news']
few_shot = ['trec', 'triviaqa', 'samsum']
synthetic = ['passage_count', 'passage_retrieval_en']
code = ['lcc', 'repobench-p']

task_types = ['Single Docs', 'Multiple Docs', 'Summarization', 'Few Shot Learning', 'Synthetic', 'Code Completion']

models = ['Qwen2-0.5B-Instruct_origin.jsonl', '']

collections = [single, multiple, summary, few_shot, synthetic, code]

models =['origin', 'inf-llm-16', 'inf-llm-24', 'retriever-only', 'retrv-inf']

for task_type, datasets in zip(task_types, collections):
    print(f"===== Task Type: {task_type} =====")

    header = "model         " + "".join([f"\t|{data:10}" for data in datasets])
    outstr = header + "\n" + "=" * 70 + "\n"
    for model in models:
        outstr += f"{model:15}" + "".join([f"\t| {all_result[model][data]:05.2f} " for data in datasets]) + "\n"
    print(outstr)
```
